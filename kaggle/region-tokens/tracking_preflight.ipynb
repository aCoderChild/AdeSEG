# Saved-version tracking preflight
CPU-only diagnostic: verify kernel startup, attached Kaggle secret and access to the existing Forge W&B project before allocating a GPU. No scientific results or model run are produced. Reuses scripts/verify_wandb_connection.py at a fixed revision.

```python
print('[preflight] kernel_started', flush=True)
import json, os, sys, urllib.request
from pathlib import Path
from kaggle_secrets import UserSecretsClient
revision = 'c072d40'
print('[preflight] checking attached WANDB_API_KEY (value never printed)', flush=True)
secret = UserSecretsClient().get_secret('WANDB_API_KEY')
if not secret: raise RuntimeError('WANDB_API_KEY secret is empty')
os.environ.update(WANDB_API_KEY=secret, WANDB_BASE_URL='https://forge.coreweave.com/api/wandb', WANDB_PROJECT='ade-seg', WANDB_ENTITY='adenoid-hypertrophy')
del secret
print('[preflight] secret accessible; fetching pinned connectivity check', flush=True)
url = 'https://raw.githubusercontent.com/aCoderChild/AdeSEG/' + revision + '/scripts/verify_wandb_connection.py'
with urllib.request.urlopen(url, timeout=60) as response: source = response.read().decode('utf-8')
sys.argv = ['verify_wandb_connection.py', '--check-only']
exec(compile(source, 'verify_wandb_connection.py', 'exec'), {'__name__': '__main__'})
result = {'state': 'complete', 'stage': 'tracking_preflight', 'revision': revision, 'secret_accessible': True, 'wandb_project_accessible': True, 'scientific_results': False, 'full_study_complete': False}
Path('/kaggle/working/tracking_preflight.json').write_text(json.dumps(result, indent=2) + '\n')
print('[preflight] complete', json.dumps(result), flush=True)
```
